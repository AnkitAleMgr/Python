```python
# importing library
import pandas as pd
import numpy as np
```

### first option for crating dataframe:

##### By using array form numpy

```python
# crating dataframe using array from numpy:
data = np.array([[1,2], [2,3], [4,5]])
```

```python
df = pd.DataFrame(data, index = ['row1', 'row2', 'row3'], columns = ['column1', 'column2'])
```

```python
df
```

### Second option for crating dataframe:

##### By making multi dimenson list

```python
# crating list
data = [[1,2], [2,3], [3,4], [4,5]]
```

```python
df = pd.DataFrame(data)
```

```python
df
```

### Third option for crating dataframe:

##### By making Dictonary

```python
# crating list for dictonary:
value_1 = ["Ankit", "Anmol", "Sujal", "Dipesh"]
value_2 = [20, 30, 40, 50]

# creating dictionary option 1
data = {"Name": value_1 , "Number" : value_2}

# creating dictionary option 2
# data = {"Name": ["ankit", "anmol"] , "Number" : [20, 30]}
```

```python
df = pd.DataFrame(data)
```
