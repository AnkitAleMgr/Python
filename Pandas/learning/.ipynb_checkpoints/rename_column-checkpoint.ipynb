```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("csv_files/StudentsPerformance.csv")
```

### Rename

##### it usuage:
1. renaming column
2. renaming index like index 0 to A, 1 to b ....
   

```python
clone_df = df.copy()
clone_df.head()
```

#### Renaming cloumn using rename method

```python
## renaming using method
clone_df.rename(columns= {"gender" : "Gender", "race/ethnicity": "Race"}).head(3)
```

#### renaming index using rename with index parameters

```python
clone_df.rename(index= {0 : "A", 1 : "B", 3: "C"}).head(5)
```
